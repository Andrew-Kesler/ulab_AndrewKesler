What is the difference between artificial intelligence, machine learning and deep learning?

```python
# AI is an umbrella term, used to describe computers that perform tasks like human 
# intelligence (including recognize patterns, make decisions, and learn from experience.

# Machine Learning (ML) is a subset of AI, where computers learn from data rather than
# being programmed explicitly.

# Deep Learning (DL) is a subset of ML, which uses neural networks that have multiple 
# layers to process the data. It is good at understanding complex patterns, and can
# map inputs to outputs with neurons.
```

What is a neural network?

```python
# A neural network is a model inspired by how the human brain works in order to 
# learn/understand patterns. It has 3 layers: the input, hidden, and output layers.
```

In lecture, we used `torch.nn.Linear(2, 50)`. What do the numbers `2` and `50` represent? 

```python
# In torch.nn.Linear(2, 50), the number 2 represents the input layer and 50 
# represents the hidden layer.
```

What does the activation function do?

```python
# An activation funtion looks like torch.nn.Sigmoid() and it introduces 
# non-linearity into the model. It helps the neurons work in a smoother and 
# more continuous way.
```

What is a loss function? Why do we need one?

```python
# A loss fucntion looks like loss_fn = torch.nn.MSELoss() and it tells us how wrong 
# our model is. It is important to have one so that the model can learn how off it is and 
# optimize the loss with an optimizer.
```

Why do we call `optimizer.zero_grad()` before calling `loss.backward()`?

```python
# We call optimizer.zero_grad() to reset gradients and then call loss.backward() to 
# re-compute gradients. This way, each iteration runs smoothly.
```

```python
import torch
import data  

# Define a new neural network
model = torch.nn.Sequential(
    torch.nn.Linear(2, 64),
    torch.nn.ReLU(),
    torch.nn.Linear(64, 32),
    torch.nn.ReLU(),
    torch.nn.Linear(32, 1)
)

# Define loss function and optimizer
loss_fn = torch.nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)

# Train the model
for t in range(10001):  # Train for 10,000 steps
    y_pred = model(data.x)
    loss = loss_fn(y_pred, data.y)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    
    if t % 1000 == 0:
        print(f"Iteration {t}, Loss: {loss.item()}")
```
